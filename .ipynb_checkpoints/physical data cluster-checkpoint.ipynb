```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE,MDS, Isomap,LocallyLinearEmbedding,SpectralEmbedding
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import fcluster
```

## 记录信息

```python
# -----------------------读取连续谱和线强度信息--------------------------------

path_line = 'C:\\Users\\zyx\\Desktop\\Spectral with Machine Learning\\data\\Continuum and Line intensities.txt'

df_line = pd.read_csv(path_line)
print(df_line.head())
print(df_line['Source'][0][0:4])
```

```python
# ------------------------记录源的谱线信息（用来聚类）----------------------------------

sources_features = df_line.drop(columns=['Source']).to_numpy()   # 只保留数值列
print("特征形状:", sources_features.shape)

sources_name = df_line['Source'].to_numpy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(sources_features)

print(sources_name)

# plt.plot(sources_features)
# plt.show()

# plt.plot(X_scaled)
# plt.show()
```

## 聚类

### PCA

```python
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
pca.explained_variance_ratio_
```

```python
plt.figure(figsize=(18,12))
for (x,y,z),src in zip(X_pca, sources_name):
    plt.scatter(x,y,s=200)
    plt.annotate(src, (x, y), fontsize=20, alpha=0.7)
    
plt.grid()
plt.show()
```

### 层次聚类

```python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
# ------------------用原始数据聚类---------------------------
# 绘制树状图帮助确定切割位置
linked = linkage(X_scaled, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(linked, labels=sources_name, orientation='top', distance_sort='descending')
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Source')
plt.ylabel('Distance')
plt.show()
```

```python
# --------------------用PCA的结果聚类------------------------------
linked_pca = linkage(X_pca, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(linked_pca, labels=sources_name, orientation='top', distance_sort='descending')

plt.title('Hierarchical Clustering Dendrogram(PCA)')
plt.xlabel('Source')
plt.ylabel('Distance')
plt.show()
```

```python


# 切割
labels_pca = fcluster(linked_pca, t=6, criterion='distance')
print(labels_pca)
print(len(labels_pca))

labels = fcluster(linked, t=6, criterion='distance')
print(labels)
print(len(labels))

# ------------------------把标签加回原始df表格-----------------------------------
df_line['cluster'] = labels
df_line['cluster_pca'] = labels_pca
```

```python
# ----------------------在PCA上展示两种聚类结果---------------------------
fig, axes = plt.subplots(1, 2,figsize=(10, 5))  
axes[0].scatter(X_pca[:,0], X_pca[:,1], c=labels)
axes[1].scatter(X_pca[:,0], X_pca[:,1], c=labels_pca)
axes[0].set_title('cluster')
axes[1].set_title('cluster(PCA)')
```

```python
# ---------------------观察各类每个维度的平均值------------------------
#target_cols = ['continuum', 'SO2', 'CH3OH']

cluster_mean = df_line.groupby('cluster').mean()
cluster_mean_pca = df_line.groupby('cluster_pca').mean()
```

### 剔除掉探测率<10%的分子

```python
# 过滤掉 Source Name, ID 等字符串列
df_numeric = df_line1.select_dtypes(include=['number'])


n_sources = df_numeric.shape[0] 
threshold = 0.1 * n_sources   


detection_counts = (df_numeric > 0).sum()

# 找出满足条件的列名
keep_columns = detection_counts[detection_counts >= threshold].index.tolist()

# 打印一下筛选结果，看看哪些分子被剔除了
removed_columns = [c for c in df_line1.columns if c not in keep_columns]
print(f"保留的分子数: {len(keep_columns)}")
print(f"剔除的分子 (探测率 < 20%): {removed_columns}")

# ------------------------------------ 筛选后的数据------------------------------------
df_filtered = df_line1[keep_columns]

# 重新标准化（非常重要！剔除列后必须重新标准化）
scaler = StandardScaler()
data_scaled_filtered = scaler.fit_transform(df_filtered)

# 重新运行 PCA
pca_new = PCA(n_components=2)
pca_new.fit(data_scaled_filtered)

# 重新获取 Loading 数据
loadings_new = pca_new.components_.T * np.sqrt(pca_new.explained_variance_)

# --- 3. 准备绘图用的索引 ---
# 既然已经筛选过了，我们直接从 keep_columns 里找你关心的分子
selected_molecules = ['34SO2','CH3OCH3','CH3OH','CO',
                      'H13CO+','HC15N','HCO+','SO','SO2', 
                      'SiO', 't-HCOOH'] 

# 只有既在“保留名单”里，又在“你关心的名单”里的分子才会被画出来
final_selection = [m for m in keep_columns if any(m.startswith(base) for base in selected_molecules)]
indices_new = [keep_columns.index(m) for m in final_selection]

load_x_new = loadings_new[indices_new, 0]
load_y_new = loadings_new[indices_new, 1]
selected_names = [df_filtered.columns[i] for i in indices_new]

# --- 2. 绘图 ---
plt.figure(figsize=(10, 8))

# 画代表分子的点 (Points for molecules)
plt.scatter(load_x_new, load_y_new, color='red', s=100, zorder=3)

# 绘制从原点到点的辅助箭头 (也可以注释掉，只保留点)
for x, y in zip(load_x_new, load_y_new):
    plt.arrow(0, 0, x, y, color='red', alpha=0.2, head_width=0.02, zorder=2)

# 标注分子名称
for i, txt in enumerate(selected_names):
    plt.text(load_x_new[i] * 1.05, load_y_new[i] * 1.05, txt, 
             fontsize=12, fontweight='bold', ha='center', va='center')

# # --- 3. 装饰图表 ---
# # 画圆圈：展示相关性强度（Loadings 都在 -1 到 1 之间）
# circle = plt.Circle((0, 0), 1, color='blue', fill=False, linestyle='--', alpha=0.3)
# plt.gca().add_artist(circle)

# plt.axhline(0, color='black', linestyle='-', linewidth=1, alpha=0.5)
# plt.axvline(0, color='black', linestyle='-', linewidth=1, alpha=0.5)
# plt.xlabel(f'PC1 ({pca1.explained_variance_ratio_[0]*100:.1f}%)', fontsize=14)
# plt.ylabel(f'PC2 ({pca1.explained_variance_ratio_[1]*100:.1f}%)', fontsize=14)
# plt.title('Molecular Loading Plot (Variables Space)', fontsize=16)
# plt.grid(alpha=0.2)
# plt.axis('equal') 
# plt.xlim(-1.1, 1.1) # Loading 最大值为 1
# plt.ylim(-1.1, 1.1)
plt.show()
```

### score plot

```python
# --- 1. 计算 Score (30个源在 PCA 空间的坐标) ---
# 使用你刚才运行 pca_new.fit() 时的那个 data_scaled_filtered
scores = pca_new.transform(data_scaled_filtered)

# --- 2. 准备绘图 ---
plt.figure(figsize=(10, 8))

# 建议：找一个你已知的物理量给点上色，比如旋转温度 Trot 或 L/M
# 如果没有，可以直接用固定颜色 c='blue'
# color_values = df_line1['Trot'] # 假设你有这一列

sc = plt.scatter(scores[:, 0], scores[:, 1], 
                 s=100, alpha=0.7, edgecolors='k',
                 c='blue', cmap='viridis') # 如果有物理量，c=color_values

# --- 3. 自动标注每个点的名字 (Source Name) ---
# 假设 df_line1 的 index 或者是某一列是源的名字
source_names = df_line1['Source'] # 或者使用特定的列名

for i, name in enumerate(source_names):
    plt.text(scores[i, 0] + 0.05, scores[i, 1] + 0.05, name, fontsize=9)

# --- 4. 装饰图表 ---
plt.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.5)
plt.axvline(0, color='black', linestyle='--', linewidth=1, alpha=0.5)

# 坐标轴标签要与 Loading Plot 对应
plt.xlabel(f'PC1 ({pca_new.explained_variance_ratio_[0]*100:.1f}%)', fontsize=12)
plt.ylabel(f'PC2 ({pca_new.explained_variance_ratio_[1]*100:.1f}%)', fontsize=12)
plt.title('PCA Score Plot (Distribution of 30 Sources)', fontsize=14)
plt.grid(alpha=0.2)
plt.show()
```

```python
print(scores.shape)
print(len(source_names))

df_pca = pd.DataFrame({
    'source': source_names,
    'PC1': scores[:, 0],
    'PC2': scores[:, 1],
})

#df_pca.to_csv('C:\\Users\\zyx\\Desktop\\Spectral with Machine Learning\\data\\pca_scores.txt', index=False)
```

## 在原有数据加上新数据

```python
# 指定需要从 df2 加入的列名
cols_to_add = ['CO','HCO+'] 

# 1. 提取基础源名并判断后缀是否为 "-1"
df_line['base_source'] = df_line['Source'].str.split('-').str[0]
df_line['is_minus1'] = df_line['Source'].str.endswith('-1')

# 2. 左连接，只选取需要的列
df_merged = df_line.merge(
    df_line1[['source'] + cols_to_add],
    left_on='base_source',
    right_on='source',
    how='left',
    suffixes=('', '_from_df2')
)

# 3. 对于非 "-1" 后缀的行，将这些新加入的列置为 0
for col in cols_to_add:
    # 处理可能的重名问题
    new_col = col if col not in df_line.columns else f"{col}_from_df2"
    df_merged.loc[~df_merged['is_minus1'], new_col] = 0

# 4. 删除辅助列
df_result = df_merged.drop(columns=['base_source', 'is_minus1', 'source'])

print(df_result)
```

```python
sources_features_merge = df_result.drop(columns=['Source']).to_numpy()   # 只保留数值列
print("特征形状:", sources_features_merge.shape)

sources_name_merge = df_result['Source'].to_numpy()

scaler_merge = StandardScaler()
X_scaled_merge = scaler_merge.fit_transform(sources_features_merge)

print(sources_name_merge)

pca_merge = PCA(n_components=3)
X_pca_merge = pca_merge.fit_transform(X_scaled_merge)
pca_merge.explained_variance_ratio_

plt.figure(figsize=(18,12))
for (x,y,z),src in zip(X_pca_merge, sources_name_merge):
    plt.scatter(x,y,s=200)
    plt.annotate(src, (x, y), fontsize=20, alpha=0.7)
    
plt.grid()
plt.show()
```

```python
# 绘制树状图帮助确定切割位置
linked_merge = linkage(X_scaled_merge, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(linked_merge, labels=sources_name_merge, orientation='top', distance_sort='descending')
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Source')
plt.ylabel('Distance')
plt.show()
```

```python
# --------------------用PCA的结果聚类------------------------------
linked_pca_merge = linkage(X_pca_merge, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(linked_pca_merge, labels=sources_name_merge, orientation='top', distance_sort='descending')

plt.title('Hierarchical Clustering Dendrogram(PCA)')
plt.xlabel('Source')
plt.ylabel('Distance')
plt.show()
```

```python
# 切割
labels_pca1_merge = fcluster(linked_pca_merge, t=10, criterion='distance')
print(labels_pca1_merge)
print(len(labels_pca1_merge))

labels_merge = fcluster(linked_merge, t=10, criterion='distance')
# print(labels1)
# print(len(labels1))

# # ------------------------把标签加回原始df表格-----------------------------------
# df_result['cluster'] = labels_merge
# df_result['cluster_pca'] = labels_pca1_merge


# ----------------------在PCA上展示两种聚类结果---------------------------
fig, axes = plt.subplots(1, 2,figsize=(10, 5))  
axes[0].scatter(X_pca_merge[:,0], X_pca_merge[:,1], c=labels_merge)
axes[1].scatter(X_pca_merge[:,0], X_pca_merge[:,1], c=labels_pca1_merge)
axes[0].set_title('cluster')
axes[1].set_title('cluster(PCA)')
```
